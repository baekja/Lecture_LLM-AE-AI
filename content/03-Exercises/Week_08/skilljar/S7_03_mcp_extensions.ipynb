## 1. MCP와 Claude Code의 관계

MCP (Model Context Protocol)는 Claude Code에 **새로운 도구, 데이터 소스, 프롬프트**를 추가합니다.

```
Claude Code (기본 능력)
├── Read, Write, Edit (파일 관리)
├── Bash (명령 실행)
├── Glob, Grep (검색)
│
└── + MCP 서버 (확장)
    ├── 파일시스템 (고급 파일 관리)
    ├── 데이터베이스 (SQL 쿼리)
    ├── GitHub (이슈, PR)
    ├── 웹 검색 (실시간 정보)
    └── 커스텀 (건축 법규 DB 등)
```

## 2. MCP 서버 설정

Claude Code의 설정 파일 (`.claude/settings.json`)에서 MCP 서버를 등록합니다.

````python
import json

# MCP 서버 설정 예시
mcp_config = {
    "mcpServers": {
        "filesystem": {
            "command": "npx",
            "args": ["-y", "@anthropic-ai/mcp-filesystem", "/path/to/project"]
        },
        "postgres": {
            "command": "npx",
            "args": ["-y", "@anthropic-ai/mcp-postgres", "postgresql://localhost/mydb"]
        },
        "github": {
            "command": "npx",
            "args": ["-y", "@anthropic-ai/mcp-github"],
            "env": {
                "GITHUB_TOKEN": "ghp_..."
            }
        }
    }
}

print("=== MCP 서버 설정 예시 ===")
print(json.dumps(mcp_config, indent=2, ensure_ascii=False))
````

````python
# 건축공학 MCP 서버 설정 예시
kds_mcp_config = {
    "mcpServers": {
        "kds-regulations": {
            "command": "python",
            "args": ["-m", "kds_mcp_server"],
            "cwd": "/path/to/kds-server",
            "env": {
                "KDS_DB_PATH": "/path/to/kds.db"
            }
        },
        "structural-calc": {
            "command": "python",
            "args": ["-m", "structural_calc_server"],
            "cwd": "/path/to/calc-server"
        }
    }
}

print("=== 건축공학 MCP 서버 설정 ===")
print(json.dumps(kds_mcp_config, indent=2, ensure_ascii=False))
print()
print("이 설정을 .claude/settings.json에 추가하면")
print("Claude Code에서 KDS 기준 조회와 구조 계산을 할 수 있습니다.")
````

## 3. MCP 서버 활용 시나리오

| MCP 서버 | 활용 예시 |
| --- | --- |
| **파일시스템** | 대규모 프로젝트 파일 검색, 일괄 처리 |
| **PostgreSQL** | 설계 데이터 조회, 결과 저장 |
| **GitHub** | 이슈 관리, PR 생성, 코드 리뷰 |
| **KDS 법규** | 기준코드 조항 조회, 설계값 확인 |
| **구조 계산** | RC 부재 설계 검토, 내진설계 검토 |

## 4. Computer Use 소개

Computer Use는 Claude가 컴퓨터 화면을 보고, 마우스와 키보드를 제어하는 실험적 기능입니다.

### 건축공학 활용 가능성
- CAD/BIM 소프트웨어 자동화 (Revit, AutoCAD)
- 구조해석 프로그램 조작 (Midas, ETABS)
- 반복적인 도면 작업 자동화
- 데이터 입력 자동화

````python
# Computer Use API 구조 (참고용 - 실제 실행에는 화면 환경 필요)
computer_use_tool = {
    "type": "computer_20250124",
    "name": "computer",
    "display_width_px": 1920,
    "display_height_px": 1080,
    "display_number": 0
}

print("=== Computer Use 도구 정의 ===")
print(json.dumps(computer_use_tool, indent=2))
print()
print("Computer Use는 현재 beta 기능입니다.")
print("API를 통해 활성화하며, 화면 캡처 + 마우스/키보드 제어를 지원합니다.")
````
