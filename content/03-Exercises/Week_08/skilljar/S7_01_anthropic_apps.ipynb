## 2. Claude API — 이전 주차 복습

Week 02-07에서 학습한 API 호출 방식을 복습합니다.

````python
# 환경 설정
import os
from dotenv import load_dotenv

load_dotenv()

# API 키 확인
api_key = os.getenv("ANTHROPIC_API_KEY")
if api_key:
    print(f"API 키 설정됨: {api_key[:10]}...")
else:
    print("⚠️ ANTHROPIC_API_KEY가 설정되지 않았습니다.")
    print("  .env 파일에 ANTHROPIC_API_KEY=sk-ant-... 을 추가하세요.")
````

````python
from anthropic import Anthropic

client = Anthropic()
MODEL = "claude-sonnet-4-5"

# 간단한 API 호출 (복습)
response = client.messages.create(
    model=MODEL,
    max_tokens=256,
    messages=[
        {"role": "user", "content": "Claude.ai, Claude Code, Claude API의 차이를 한 문장씩 설명해줘."}
    ]
)

print(response.content[0].text)
````

## 3. 세 가지 접근 경로 비교 실습

같은 작업을 세 가지 경로로 수행할 때의 차이를 체험합니다.

### 과제: "RC 보의 최소 철근비를 계산하라"

| 경로 | 수행 방법 |
| --- | --- |
| **Claude.ai** | 웹에서 직접 질문 → 즉시 답변 |
| **Claude Code** | 터미널에서 계산 함수 생성 + 테스트 실행 |
| **Claude API** | Python 코드에서 API 호출 → 프로그래밍 방식 처리 |

````python
# API 경로: 프로그래밍 방식으로 구조 계산
response = client.messages.create(
    model=MODEL,
    max_tokens=512,
    system="당신은 건축구조 전문가입니다. KDS 14 20 20 기준을 따릅니다.",
    messages=[
        {"role": "user", "content": 
         "RC 보의 최소 철근비를 계산해줘.\n"
         "조건: fck=27MPa, fy=400MPa\n"
         "KDS 14 20 20 기준으로 계산하고, 수식을 포함해줘."}
    ]
)

print("=== API 경로 결과 ===")
print(response.content[0].text)
````

## 4. 선택 기준 정리

### 의사결정 플로우차트

```
코딩이 필요한가?
├── 아니오 → Claude.ai (웹에서 대화)
└── 예 → 프로덕션 배포?
    ├── 예 → Claude API (SDK로 통합)
    └── 아니오 → 코드 프로젝트 작업?
        ├── 예 → Claude Code (터미널에서 개발)
        └── 아니오 → Claude.ai
```

### 핵심 포인트
- **Claude.ai**: 빠른 프로토타이핑, 아이디어 검증
- **Claude Code**: 코드 프로젝트 개발, 파일 시스템 접근
- **Claude API**: 프로덕션 시스템, 자동화 파이프라인
- **실무**: 세 가지를 단계별로 조합하여 사용
