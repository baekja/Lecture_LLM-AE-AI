## 1. Claude Code 설치

### 시스템 요구사항
- Node.js 18 이상
- macOS, Linux, Windows (WSL2)

### 설치 명령어
```bash
# Node.js 버전 확인
node --version

# Claude Code 설치
npm install -g @anthropic-ai/claude-code

# 설치 확인
claude --version

# 첫 실행 (프로젝트 디렉토리에서)
cd my-project/
claude
```

````python
# Node.js 설치 확인 (이 셀은 터미널 명령을 실행합니다)
import subprocess

try:
    result = subprocess.run(["node", "--version"], capture_output=True, text=True)
    version = result.stdout.strip()
    major = int(version.lstrip('v').split('.')[0])
    if major >= 18:
        print(f"Node.js {version} -- 요구사항 충족 (18 이상)")
    else:
        print(f"Node.js {version} -- 업데이트 필요 (18 이상 필요)")
except FileNotFoundError:
    print("Node.js가 설치되지 않았습니다. https://nodejs.org 에서 설치하세요.")
````

````python
# Claude Code 설치 확인
try:
    result = subprocess.run(["claude", "--version"], capture_output=True, text=True)
    if result.returncode == 0:
        print(f"Claude Code 설치됨: {result.stdout.strip()}")
    else:
        print("Claude Code가 설치되지 않았습니다.")
        print("설치: npm install -g @anthropic-ai/claude-code")
except FileNotFoundError:
    print("Claude Code가 설치되지 않았습니다.")
    print("설치: npm install -g @anthropic-ai/claude-code")
````

## 2. CLAUDE.md 작성법

CLAUDE.md는 Claude Code가 프로젝트를 이해하는 가장 중요한 문서입니다.

### CLAUDE.md 필수 구성 요소

| 섹션 | 내용 | 예시 |
| --- | --- | --- |
| 프로젝트 개요 | 프로젝트의 목적과 핵심 기능 | "건축 시방서 분석 RAG 시스템" |
| 기술 스택 | 사용 중인 언어, 프레임워크, 라이브러리 | Python 3.11, FastAPI, ChromaDB |
| 디렉토리 구조 | 주요 디렉토리와 역할 | src/api/, src/rag/, tests/ |
| 빌드/실행 | 설치, 실행, 테스트 명령어 | pip install, uvicorn, pytest |
| 코딩 컨벤션 | 스타일 규칙, 네이밍, 문서화 | 타입 힌트 필수, Google docstring |

````python
# CLAUDE.md 템플릿 생성기
def generate_claude_md_template(
    project_name: str,
    description: str,
    tech_stack: list[str],
    directories: dict[str, str],
    commands: dict[str, str]
) -> str:
    """건축공학 프로젝트용 CLAUDE.md 템플릿을 생성합니다."""
    
    template = f"""# CLAUDE.md

## 프로젝트 개요
{description}

## 기술 스택
"""
    for tech in tech_stack:
        template += f"- {tech}\n"
    
    template += "\n## 디렉토리 구조\n"
    for dir_name, role in directories.items():
        template += f"{dir_name:<20} {role}\n"
    
    template += "\n## 빌드 및 실행\n"
    for cmd_name, cmd in commands.items():
        template += f"{cmd:<40} # {cmd_name}\n"
    
    template += """
## 도메인 규칙
### 단위 시스템
- 힘: kN, 모멘트: kN·m, 응력: MPa
- 길이: mm (계산), m (도면)

### 코딩 컨벤션
- 타입 힌트 필수
- Docstring: Google 스타일
- 변수명: 기준코드 표기 (fck, fy, As 등)
"""
    return template


# 예시: 건축 구조 검토 시스템
claude_md = generate_claude_md_template(
    project_name="structural-review",
    description="KDS 기준 RC 부재 구조 설계 검토 자동화 시스템",
    tech_stack=["Python 3.11+", "FastAPI", "Streamlit", "ChromaDB", "Anthropic Claude API"],
    directories={
        "src/api/": "FastAPI 라우터",
        "src/calculators/": "구조 계산 모듈",
        "src/rag/": "RAG 파이프라인",
        "src/ui/": "Streamlit 페이지",
        "tests/": "pytest 테스트",
    },
    commands={
        "설치": "pip install -r requirements.txt",
        "백엔드": "uvicorn src.api.main:app --reload",
        "프론트엔드": "streamlit run src/ui/app.py",
        "테스트": "pytest -v",
    }
)

print(claude_md)
````

## 3. 슬래시 명령어

Claude Code 세션 내에서 사용하는 제어 명령어:

| 명령어 | 설명 | 사용 시점 |
| --- | --- | --- |
| `/init` | CLAUDE.md 자동 생성 | 새 프로젝트 시작 시 |
| `/compact` | 대화 요약 (컨텍스트 절약) | 긴 대화에서 메모리 부족 시 |
| `/clear` | 대화 초기화 | 새 작업 시작 시 |
| `/cost` | 토큰 사용량 표시 | 비용 관리 시 |
| `/model` | 모델 변경 | 모델 전환 시 |
| `/help` | 도움말 | 도움 필요 시 |

### 실습: 터미널에서 Claude Code 실행하기

```bash
# 1. 프로젝트 디렉토리로 이동
cd ~/my-structural-project/

# 2. Claude Code 시작
claude

# 3. /init으로 프로젝트 분석
> /init

# 4. CLAUDE.md 확인 및 수정
> CLAUDE.md를 읽고 건축공학 도메인 정보를 추가해줘.

# 5. 비용 확인
> /cost
```

## 4. 권한 시스템

### 권한 수준
- **자동 허용**: Read, Glob, Grep (안전한 읽기 작업)
- **확인 필요**: Write, Edit, Bash (파일 변경/명령 실행)
- **항상 확인**: rm, git push, sudo (위험한 작업)

### allowedTools 설정
```bash
# 자주 사용하는 안전한 도구를 사전 허용
claude config set allowedTools "Edit,Write,Bash(git *)"
```
