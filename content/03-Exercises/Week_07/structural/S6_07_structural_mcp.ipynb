## §3. 첫 번째 통합 — 휨 강도 도구와 전단 강도 도구 정의

한국 콘크리트구조 설계기준에 따른 콘크리트 보의 두 가지 핵심 검토 도구를 정의합니다. 이 두 도구는 본 트랙에서 가장 자주 호출되는 핵심 자산이며, 한국 설계기준 인용 검토 시나리오의 중심에 놓입니다. 기존 백업 파일의 셀 세 개를 그대로 인용하여 재현합니다.

첫 번째 도구인 **휨 강도 검토 함수** 는 한국 콘크리트구조 설계기준의 휨 부재 조문에 근거합니다. 등가 응력블록 깊이를 산정한 뒤 공칭 휨강도를 구하고, 마지막으로 강도감소계수 영점 팔오를 곱하여 설계 휨강도를 산정합니다. 동시에 최소 인장철근비 식을 만족하는지 함께 검토하여 종합적인 휨 안전성을 평가합니다. 등가 응력블록 깊이는 인장 철근 단면적과 철근 항복강도의 곱을 영점 팔오와 콘크리트 강도와 보 폭의 곱으로 나눈 값으로 산정됩니다. 공칭 휨강도는 인장 철근 단면적과 철근 항복강도와 유효 깊이에서 등가 응력블록 깊이의 절반을 뺀 값을 모두 곱한 결과입니다.

두 번째 도구인 **전단 강도 검토 함수** 는 한국 콘크리트구조 설계기준의 전단 부재 조문에 근거합니다. 콘크리트 분담 전단강도와 전단보강근 분담 전단강도를 더한 뒤 강도감소계수 영점 칠오를 곱하여 설계 전단강도를 산정합니다. 콘크리트 분담은 콘크리트 강도의 제곱근에 보 폭과 유효 깊이를 곱한 뒤 일정 계수를 적용해 산정합니다. 전단보강근 분담은 보강근 단면적과 항복강도와 유효 깊이의 곱을 보강근 간격으로 나눈 값으로 산정됩니다.


````python
# 기존 백업 S6_07.ipynb cell 3 인용 — 휨/전단 도구
@mcp.tool()
def check_flexural_strength(
    b: float = Field(description="보 폭 (mm)"),
    d: float = Field(description="유효 깊이 (mm)"),
    As: float = Field(description="인장 철근 단면적 (mm²)"),
    fck: float = Field(description="콘크리트 fck (MPa)"),
    fy: float = Field(description="철근 fy (MPa)"),
    Mu: float = Field(description="소요 휨모멘트 (kN·m)"),
) -> str:
    """RC 보 휨강도 검토 (KDS 41 17 00, φ=0.85)."""
    a = As * fy / (0.85 * fck * b)
    Mn = As * fy * (d - a / 2) / 1e6  # kN·m
    phi = 0.85
    phi_Mn = phi * Mn
    rho = As / (b * d)
    rho_min = max(0.25 * math.sqrt(fck) / fy, 1.4 / fy)
    return json.dumps({
        "section": f"{b}x{d}",
        "a_mm": round(a, 2),
        "phi_Mn_kNm": round(phi_Mn, 2),
        "Mu_kNm": Mu,
        "DCR": round(Mu / phi_Mn, 3) if phi_Mn > 0 else None,
        "flexure_check": "OK" if phi_Mn >= Mu else "NG",
        "rho": round(rho, 5),
        "rho_min": round(rho_min, 5),
        "rho_check": "OK" if rho >= rho_min else "NG",
        "reference": "KDS 41 17 00 §4.3",
    }, indent=2, ensure_ascii=False)


@mcp.tool()
def check_shear_strength(
    b: float = Field(description="보 폭 (mm)"),
    d: float = Field(description="유효 깊이 (mm)"),
    fck: float = Field(description="콘크리트 fck (MPa)"),
    Av: float = Field(description="전단보강근 단면적 (mm²)"),
    s: float = Field(description="전단보강근 간격 (mm)"),
    fy: float = Field(description="전단보강근 fy (MPa)"),
    Vu: float = Field(description="소요 전단력 (kN)"),
) -> str:
    """RC 보 전단강도 검토 (KDS 41 17 00, φ=0.75)."""
    Vc = (1 / 6) * math.sqrt(fck) * b * d / 1000.0
    Vs = Av * fy * d / s / 1000.0
    Vn = Vc + Vs
    phi = 0.75
    phi_Vn = phi * Vn
    return json.dumps({
        "Vc_kN": round(Vc, 2),
        "Vs_kN": round(Vs, 2),
        "phi_Vn_kN": round(phi_Vn, 2),
        "Vu_kN": Vu,
        "DCR": round(Vu / phi_Vn, 3) if phi_Vn > 0 else None,
        "shear_check": "OK" if phi_Vn >= Vu else "NG",
        "reference": "KDS 41 17 00 §4.4",
    }, indent=2, ensure_ascii=False)


print("✅ v1 도구 2개 등록")
````

### 첫 번째 단계 시연 — 폭 삼백 밀리미터, 유효깊이 오백사십 밀리미터 보 사례

방금 정의한 두 도구를 실제 단면으로 호출하여 결과가 합리적인지 확인합니다. 입력 조건은 다음과 같이 설정합니다.

## §5. 세 번째 통합 — 다섯 번째 주차의 검색 보강 생성 체인을 동적 리소스로 노출

세 번째 단계 노트북에서 정의한 빈도 기반 한국 설계기준 검색기와 두 가지 동적 리소스를 본 통합본에 옮겨 옵니다. 작업 순서를 정리하면 다음과 같습니다. 먼저 한국 설계기준 코퍼스를 정의하여 본 통합본 안에서 자가완결적으로 검색이 동작하도록 합니다. 다음으로 빈도 기반 알고리즘을 구현한 검색 진입점 함수를 만듭니다. 그러고 나서 자유 검색 리소스와 조문 식별자 기반 검색 리소스 두 가지를 등록합니다.

이로써 거대언어모델은 두 가지 방식으로 한국 설계기준 조항을 인용할 수 있게 됩니다. 첫째 방식은 정적 요약 리소스를 통해 한국 설계기준 전체 그림을 잡는 것입니다. 둘째 방식은 동적 검색 리소스를 통해 구체적인 조항을 찾아 정확한 인용을 만들어 내는 것입니다. 이 두 방식은 상호 보완적으로 작동하여, 답변의 큰 그림은 요약에서 가져오고 세부 인용은 검색에서 가져오는 식으로 자연스럽게 결합됩니다.


````python
# S6_st03 인용 — KDS RAG 코퍼스 + BM25 + 리소스 2개
KDS_SECTIONS = [
    {"id": "KDS-41-10-15-010", "text": "콘크리트 fck 최소 21 MPa 이상."},
    {"id": "KDS-41-10-15-020", "text": "이형철근 fy SD400 또는 SD500."},
    {"id": "KDS-41-10-20-010", "text": "RC 보 최소 인장철근비 max(0.25*sqrt(fck)/fy, 1.4/fy)."},
    {"id": "KDS-41-10-20-020", "text": "전단보강근 간격 d/2 이하."},
    {"id": "KDS-41-10-20-030", "text": "기둥 최소 단면치수 300mm 이상."},
    {"id": "KDS-41-12-00-010", "text": "하중조합 1.2D+1.6L (기본), 1.2D+1.0L+1.0E (지진)."},
]


def _tok(text):
    return [t for t in re.split(r"\W+", text.lower()) if t]


_DOC_TOKS = [_tok(s["text"]) for s in KDS_SECTIONS]
_DOC_LENS = [len(t) for t in _DOC_TOKS]
_AVG_LEN = sum(_DOC_LENS) / max(len(_DOC_LENS), 1)
_DF = Counter()
for toks in _DOC_TOKS:
    for t in set(toks):
        _DF[t] += 1


def hybrid_retrieve(query: str, k: int = 3) -> list[dict]:
    """W05 KDS RAG fallback (BM25 only)."""
    n = len(KDS_SECTIONS)
    qt = _tok(query)
    k1, b = 1.5, 0.75
    scored = []
    for i, toks in enumerate(_DOC_TOKS):
        tf = Counter(toks)
        score = 0.0
        for t in qt:
            f = tf.get(t, 0)
            df = _DF.get(t, 0)
            idf = math.log((n - df + 0.5) / (df + 0.5) + 1)
            norm = k1 * (1 - b + b * _DOC_LENS[i] / _AVG_LEN)
            score += idf * f * (k1 + 1) / (f + norm + 1e-9)
        scored.append((KDS_SECTIONS[i]["id"], KDS_SECTIONS[i]["text"], score))
    scored.sort(key=lambda x: -x[2])
    return [{"id": i, "text": t, "score": round(s, 3)} for i, t, s in scored[:k] if s > 0]


@mcp.resource("kds://search/{query}", mime_type="application/json")
def search_kds(query: str) -> str:
    return json.dumps({"query": query, "results": hybrid_retrieve(query, k=3)},
                      ensure_ascii=False, indent=2)


@mcp.resource("kds://41-17-00/{section_id}", mime_type="text/plain")
def get_kds_section(section_id: str) -> str:
    results = hybrid_retrieve(f"KDS 41 10 20 {section_id}", k=1)
    if not results:
        return f"Section {section_id}: no match"
    return f"[{results[0]['id']}] {results[0]['text']}"


print("✅ v3 RAG 리소스 2개 등록 (kds://search/*, kds://41-17-00/{id})")
````

## §6. 4차 통합 — Midas Civil `.mgt` 입력 파일 파서 도구

S6_st04 단계 노트북에서 자세히 다루는 Midas Civil의 텍스트 입력 형식 `.mgt` 파일 파서를 본 통합본에 간략 버전으로 추가합니다. 이 도구는 구조해석 입력 데이터(절점 좌표, 부재 연결 정보)를 LLM이 직접 읽을 수 있는 JSON으로 변환합니다.

`.mgt` 파일은 텍스트 기반이며, `*NODE` 섹션과 `*ELEMENT` 섹션이 별표로 구분된 키워드 다음에 콤마로 구분된 데이터 행으로 이어지는 구조를 가집니다. 본 도구는 이를 정규식 없이 라인 단위로 파싱하여 절점 좌표와 부재 연결 정보를 추출합니다. 본격적인 운영용 파서는 S6_st04에서 더 정교하게 다룹니다.


````python
# S6_st04 미리보기 — 간단 .mgt 파서
@mcp.tool()
def parse_midas_mgt(
    file_path: str = Field(description="Midas .mgt 파일 절대경로"),
) -> str:
    """Midas Civil .mgt 입력 파일에서 절점/부재를 추출합니다.

    *NODE / *ELEMENT 섹션을 정규식으로 파싱.
    실제 운영용 파서는 S6_st04 참조.
    """
    p = Path(file_path)
    if not p.exists():
        return json.dumps({"error": f"File not found: {file_path}"}, ensure_ascii=False)

    text = p.read_text(encoding="utf-8", errors="ignore")

    nodes = []
    elements = []
    section = None
    for line in text.splitlines():
        ls = line.strip()
        if not ls or ls.startswith(";"):
            continue
        if ls.startswith("*NODE"):
            section = "NODE"; continue
        if ls.startswith("*ELEMENT"):
            section = "ELEMENT"; continue
        if ls.startswith("*"):
            section = None; continue

        parts = [p.strip() for p in ls.split(",")]
        if section == "NODE" and len(parts) >= 4:
            try:
                nodes.append({"id": int(parts[0]),
                              "x": float(parts[1]),
                              "y": float(parts[2]),
                              "z": float(parts[3])})
            except ValueError:
                pass
        elif section == "ELEMENT" and len(parts) >= 5:
            try:
                elements.append({"id": int(parts[0]),
                                 "type": parts[1],
                                 "n1": int(parts[3]),
                                 "n2": int(parts[4])})
            except (ValueError, IndexError):
                pass

    return json.dumps({
        "file": file_path,
        "n_nodes": len(nodes),
        "n_elements": len(elements),
        "nodes_sample": nodes[:3],
        "elements_sample": elements[:3],
    }, indent=2, ensure_ascii=False)


print("✅ v4 도구 1개 추가: parse_midas_mgt")
````

## §11. 파이썬 모듈 파일 최종본 저장하기

지금까지 통합본 안에서 정의한 첫 번째 버전부터 다섯 번째 버전까지의 모든 자산을 단일 파이썬 모듈 파일로 디스크에 저장합니다. 최종 자산 구성은 도구 세 개와 리소스 다섯 개와 프롬프트 한 개입니다. 여섯 번째 버전인 시각적 검증 도구 실행과 일곱 번째 버전인 명령줄 도구 등록은 외부 명령이므로 파일 자체에는 포함되지 않으며, 노트북의 안내 셀에서만 다룹니다.

이 최종본 파일이 과제 제출이나 실제 명령줄 도구 등록에 사용되는 산출물입니다. 학생은 본 파일을 자신의 작업 디렉터리에 보관해 두었다가, 나중에 실제 명령줄 도구에 등록하거나 다른 프로젝트에서 참조 코드로 활용할 수 있습니다. 또한 이 파일을 한국 건축공학 도메인의 다른 영역으로 확장하는 출발점으로도 사용할 수 있습니다. 예를 들어 강구조 설계기준이나 내진 설계기준 등으로 도메인을 넓혀 가는 일도 같은 패턴을 따라 진행하면 됩니다.


````python
FINAL_SOURCE = '''"""structural-mcp — 통합본 (v1~v5)

자동 생성: S6_07_structural_mcp.ipynb (Stage 1~6 압축)
- Tools: check_flexural_strength, check_shear_strength, parse_midas_mgt
- Resources: kds://41-17-00/summary, data://materials/{concrete,rebar}-table,
             kds://search/{query}, kds://41-17-00/{section_id}
- Prompts: structural_review

Claude Code 등록:
  $ claude mcp add structural-mcp -- uv run structural_mcp.py
"""
import json
import math
import re
from collections import Counter
from pathlib import Path
from mcp.server.fastmcp import FastMCP
from pydantic import Field

mcp = FastMCP("StructuralMCP", log_level="ERROR")


# ── KDS 코퍼스 + BM25 (W05 RAG 인용) ──
KDS_SECTIONS = [
    {"id": "KDS-41-10-20-010", "text": "RC 보 최소 인장철근비 max(0.25*sqrt(fck)/fy, 1.4/fy)."},
    {"id": "KDS-41-10-20-020", "text": "전단보강근 간격 d/2 이하."},
    {"id": "KDS-41-10-20-030", "text": "기둥 최소 단면치수 300mm 이상."},
    {"id": "KDS-41-12-00-010", "text": "하중조합 1.2D+1.6L."},
]

def _tok(t): return [w for w in re.split(r"\\W+", t.lower()) if w]
_TOKS = [_tok(s["text"]) for s in KDS_SECTIONS]
_LENS = [len(t) for t in _TOKS]
_AVG = sum(_LENS) / max(len(_LENS), 1)
_DF = Counter(); [_DF.update(set(t)) for t in _TOKS]


def hybrid_retrieve(query, k=3):
    n = len(KDS_SECTIONS); qt = _tok(query); k1, b = 1.5, 0.75
    scored = []
    for i, toks in enumerate(_TOKS):
        tf = Counter(toks); s = 0.0
        for t in qt:
            f = tf.get(t, 0); df = _DF.get(t, 0)
            idf = math.log((n - df + 0.5) / (df + 0.5) + 1)
            norm = k1 * (1 - b + b * _LENS[i] / _AVG)
            s += idf * f * (k1 + 1) / (f + norm + 1e-9)
        scored.append((KDS_SECTIONS[i]["id"], KDS_SECTIONS[i]["text"], s))
    scored.sort(key=lambda x: -x[2])
    return [{"id": i, "text": t, "score": round(s, 3)} for i, t, s in scored[:k] if s > 0]


# ── Tools ──
@mcp.tool()
def check_flexural_strength(
    b: float = Field(description="보 폭 (mm)"),
    d: float = Field(description="유효 깊이 (mm)"),
    As: float = Field(description="인장 철근 As (mm²)"),
    fck: float = Field(description="콘크리트 fck (MPa)"),
    fy: float = Field(description="철근 fy (MPa)"),
    Mu: float = Field(description="소요 휨모멘트 (kN·m)"),
) -> str:
    """RC 보 휨강도 검토 (KDS 41 17 00, φ=0.85)."""
    a = As * fy / (0.85 * fck * b)
    Mn = As * fy * (d - a / 2) / 1e6
    phi_Mn = 0.85 * Mn
    rho = As / (b * d)
    rho_min = max(0.25 * math.sqrt(fck) / fy, 1.4 / fy)
    return json.dumps({
        "phi_Mn_kNm": round(phi_Mn, 2), "Mu_kNm": Mu,
        "DCR": round(Mu / phi_Mn, 3) if phi_Mn > 0 else None,
        "flexure_check": "OK" if phi_Mn >= Mu else "NG",
        "rho_check": "OK" if rho >= rho_min else "NG",
    }, indent=2, ensure_ascii=False)


@mcp.tool()
def check_shear_strength(
    b: float = Field(description="보 폭 (mm)"),
    d: float = Field(description="유효 깊이 (mm)"),
    fck: float = Field(description="콘크리트 fck (MPa)"),
    Av: float = Field(description="전단보강 Av (mm²)"),
    s: float = Field(description="전단보강 간격 (mm)"),
    fy: float = Field(description="전단보강 fy (MPa)"),
    Vu: float = Field(description="소요 전단력 (kN)"),
) -> str:
    """RC 보 전단강도 검토 (KDS 41 17 00, φ=0.75)."""
    Vc = (1 / 6) * math.sqrt(fck) * b * d / 1000.0
    Vs = Av * fy * d / s / 1000.0
    phi_Vn = 0.75 * (Vc + Vs)
    return json.dumps({
        "Vc_kN": round(Vc, 2), "Vs_kN": round(Vs, 2),
        "phi_Vn_kN": round(phi_Vn, 2), "Vu_kN": Vu,
        "DCR": round(Vu / phi_Vn, 3) if phi_Vn > 0 else None,
        "shear_check": "OK" if phi_Vn >= Vu else "NG",
    }, indent=2, ensure_ascii=False)


@mcp.tool()
def parse_midas_mgt(file_path: str = Field(description="Midas .mgt 파일 경로")) -> str:
    """Midas Civil .mgt 파일에서 절점/부재 추출."""
    p = Path(file_path)
    if not p.exists():
        return json.dumps({"error": f"Not found: {file_path}"}, ensure_ascii=False)
    text = p.read_text(encoding="utf-8", errors="ignore")
    nodes, elements, section = [], [], None
    for line in text.splitlines():
        ls = line.strip()
        if not ls or ls.startswith(";"): continue
        if ls.startswith("*NODE"): section = "NODE"; continue
        if ls.startswith("*ELEMENT"): section = "ELEMENT"; continue
        if ls.startswith("*"): section = None; continue
        parts = [p.strip() for p in ls.split(",")]
        try:
            if section == "NODE" and len(parts) >= 4:
                nodes.append({"id": int(parts[0]),
                              "x": float(parts[1]), "y": float(parts[2]), "z": float(parts[3])})
            elif section == "ELEMENT" and len(parts) >= 5:
                elements.append({"id": int(parts[0]), "type": parts[1],
                                 "n1": int(parts[3]), "n2": int(parts[4])})
        except (ValueError, IndexError):
            pass
    return json.dumps({"n_nodes": len(nodes), "n_elements": len(elements)},
                      indent=2, ensure_ascii=False)


# ── Resources ──
@mcp.resource("kds://41-17-00/summary", mime_type="application/json")
def get_kds_summary():
    return json.dumps({"flexure": {"phi": 0.85}, "shear": {"phi": 0.75}},
                      indent=2, ensure_ascii=False)


@mcp.resource("data://materials/concrete-table", mime_type="application/json")
def get_concrete_table():
    return json.dumps({"C24": {"fck": 24}, "C27": {"fck": 27}, "C30": {"fck": 30},
                       "C35": {"fck": 35}, "C40": {"fck": 40}},
                      indent=2, ensure_ascii=False)


@mcp.resource("data://materials/rebar-table", mime_type="application/json")
def get_rebar_table():
    return json.dumps({"D10": 71.3, "D13": 126.7, "D16": 198.6, "D19": 286.5,
                       "D22": 387.1, "D25": 506.7, "D29": 642.4, "D32": 794.2},
                      indent=2, ensure_ascii=False)


@mcp.resource("kds://search/{query}", mime_type="application/json")
def search_kds(query: str):
    return json.dumps({"query": query, "results": hybrid_retrieve(query, k=3)},
                      ensure_ascii=False, indent=2)


@mcp.resource("kds://41-17-00/{section_id}", mime_type="text/plain")
def get_kds_section(section_id: str):
    r = hybrid_retrieve(f"KDS 41 10 20 {section_id}", k=1)
    return f"[{r[0]['id']}] {r[0]['text']}" if r else f"Section {section_id}: no match"


# ── Prompts ──
@mcp.prompt()
def structural_review(member_type: str, b: float, d: float, fck: float, fy: float) -> str:
    return f"""당신은 건축구조 전문가입니다. KDS 41 17 00에 따라 RC {member_type}를 검토해주세요.
부재: {member_type}, b={b}mm, d={d}mm, fck={fck}MPa, fy={fy}MPa
1. 휨/전단 → check_flexural_strength, check_shear_strength
2. KDS 인용 → kds://41-17-00/summary, kds://search/{{검색어}}
3. 종합 판정 (DCR, OK/NG)"""


if __name__ == "__main__":
    mcp.run(transport="stdio")
'''

with open("structural_mcp.py", "w", encoding="utf-8") as f:
    f.write(FINAL_SOURCE)

print(f"✅ structural_mcp.py 통합본 저장 ({len(FINAL_SOURCE)} bytes)")
print("   - Tools 3, Resources 5, Prompts 1")
````

**시나리오 설정**: 사용자가 한국 콘크리트구조 설계기준에 따라 비일 보를 검토해 달라고 요청한 상황을 가정합니다. 단면 제원은 폭 삼백 밀리미터, 전체 깊이 육백 밀리미터이며, 인장 철근은 디이십오 철근 네 본을 배근했습니다. 재료 강도는 콘크리트의 경우 이십칠 메가파스칼, 철근의 경우 사백 메가파스칼입니다. 작용 휨모멘트는 이백 킬로뉴턴미터로 가정합니다.

**기대되는 워크플로**: 거대언어모델은 다음 여섯 단계로 작업을 수행합니다. 첫째 단계에서는 표준 검토 프롬프트를 호출하여 검토 절차를 표준화합니다. 부재 종류는 보, 폭은 삼백, 유효 깊이는 오백사십, 콘크리트 강도는 이십칠, 철근 강도는 사백으로 매개변수를 채웁니다. 둘째 단계에서는 한국 설계기준 요약 리소스를 자동으로 첨부하여 검토 기준 전반을 파악합니다. 셋째 단계에서는 철근 규격 표 리소스를 읽어 디이십오 철근 한 본의 단면적이 오백육 점 칠 제곱밀리미터임을 확인하고, 네 본을 곱하여 총 인장철근 단면적이 약 이천이십육 점 팔 제곱밀리미터임을 계산합니다. 넷째 단계에서는 휨 강도 검토 도구를 위 인자로 호출하여 소요 대 내력 비율을 계산합니다. 다섯째 단계에서는 검색 보강 생성 리소스를 호출하여 최소 철근비 관련 한국 설계기준 조항을 인용합니다. 여섯째 단계에서는 최종 답변에 정확한 조항 번호와 함께 안전 또는 위험 판정을 제시합니다.


````python
# End-to-End 시연 (Claude API 없이 도구만 직접 호출)
async def e2e_demo():
    print("=" * 60)
    print("시나리오: B1 보 검토 (300x600, D25 4개, fck=27, fy=400, Mu=200kN·m)")
    print("=" * 60)

    # Step 1: 재료 테이블 조회
    rebar = await mcp.read_resource("data://materials/rebar-table")
    rebar_data = json.loads(str(rebar)[0].text if hasattr(str(rebar), "0") else
                             [r.text for r in rebar if hasattr(r, "text")][0]
                             if isinstance(rebar, (list, tuple)) else str(rebar))
    print(f"\n[1] 재료 조회: D25 As = {rebar_data.get('D25', {}).get('As', 506.7)} mm²")

    As_total = 4 * 506.7  # 4-D25
    print(f"    4-D25 총 단면적 = {As_total} mm²")

    # Step 2: 휨 도구 호출
    print("\n[2] check_flexural_strength 도구 호출:")
    result = await mcp.call_tool("check_flexural_strength", {
        "b": 300, "d": 540, "As": As_total,
        "fck": 27, "fy": 400, "Mu": 200
    })
    print(f"    결과: {result}")

    # Step 3: KDS 인용
    print("\n[3] KDS 검색 (철근비):")
    rho_search = await mcp.read_resource("kds://search/철근비")
    print(f"    {str(rho_search)[:300]}")

    print("\n[4] 최종 답변 (Claude가 생성하면):")
    print("    'B1 보 휨강도 OK (DCR<1.0). KDS-41-10-20-010에 따라 최소 철근비 만족.'")

await e2e_demo()
````
