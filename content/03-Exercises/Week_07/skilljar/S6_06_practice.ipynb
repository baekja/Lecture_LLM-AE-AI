## §3. 도메인 선택 가이드 — 권장 세 가지 후보

학생이 직접 도메인을 정해도 좋지만, 어디서부터 시작해야 할지 막막하게 느껴진다면 아래의 세 가지 중 하나로 시작하는 것을 권장한다.

**첫 번째 권장 도메인 — 할 일 관리**:

데이터 모델은 항목 아이디를 키로 하고 본문 텍스트와 완료 여부와 마감일을 가지는 사전 형태가 자연스럽다. 도구 세 개는 할 일 추가와 할 일 완료 처리와 할 일 삭제로 구성하면 좋다. 리소스 두 개는 전체 목록 유알아이를 직접 형태로, 단일 항목 유알아이를 매개변수 형태로 정의한다. 프롬프트는 오늘 할 일을 요약하는 템플릿이 자연스럽다.

**두 번째 권장 도메인 — 레시피 컬렉션**:

데이터 모델은 레시피 아이디를 키로 하고 레시피 이름과 재료 목록과 조리 단계를 가지는 사전 형태가 적합하다. 도구는 레시피 추가와 레시피 검색과 레시피 평점 매기기 같은 식으로 구성한다. 리소스는 전체 레시피 목록과 단일 레시피 가져오기 두 개다. 프롬프트로는 식단을 추천하는 템플릿이 자연스럽다.

**세 번째 권장 도메인 — 개인 지식 베이스**:

데이터 모델은 노트 아이디를 키로 하고 제목과 본문과 태그를 가지는 사전 형태가 좋다. 도구는 노트 추가와 노트 갱신과 노트 삭제로 구성한다. 리소스는 전체 노트 목록과 단일 노트 가져오기 두 개다. 프롬프트는 주간 리뷰를 생성하는 템플릿이 어울린다.

**도메인 선택의 기준**:

세 가지 권장 도메인 중에서 어느 것을 고를지 망설여진다면, 다음 세 가지 기준을 떠올려 보자. 첫째 기준은 친숙함이다. 자기 일상에서 자주 쓰는 워크플로에 가까운 것을 고르는 것이 좋다. 둘째 기준은 자연스러운 역할 분담이다. 도구와 리소스와 프롬프트의 역할 분담이 자연스럽게 떠오르는 도메인을 고른다. 셋째 기준은 데이터 구조의 단순성이다. 너무 복잡하지 않은 1 단계 사전 정도의 도메인을 권장한다.

본 노트북의 스텁 코드는 할 일 관리 도메인을 가정하여 작성되어 있다. 다른 도메인을 선택했다면 식별자만 바꿔서 적용하면 된다.


## §4. 도구 슬롯 1 — 읽기 패턴

첫 번째 도구 슬롯은 읽기 패턴이다. 함수 이름은 자기 도메인의 항목을 읽는 의미를 담아 짓고, 단일 항목을 아이디로 조회하는 역할을 맡긴다. 부작용이 없다는 점에서 리소스와 비슷한 성격을 가지지만, 도구로 노출하면 클로드가 능동적으로 호출할 수 있다는 차이점이 있다. 즉 같은 데이터를 두 가지 경로로 노출함으로써, 사용 시나리오에 따라 적절한 경로를 골라 쓸 수 있게 만드는 셈이다.

이번 셀에서 학생이 채워야 할 부분은 다음과 같다. 함수의 본문을 적절히 작성하고, 매개변수마다 필드 설명을 의미 있게 작성한다. 의미 있는 설명이 중요한 이유는, 클로드가 그 설명을 읽고 도구를 어떻게 사용할지 판단하기 때문이다. 즉 설명의 품질이 도구 사용의 품질로 직결된다는 뜻이다. 설명이 모호하면 클로드가 도구를 잘못 사용할 가능성이 커지고, 설명이 명확하면 클로드가 정확하게 사용할 가능성이 커진다.


```python
# Week_07.md §2.6 line ~1467 — Tool 1: read 패턴

@mcp.tool(
    name='read_item',
    description='Read a single item by ID',
)
def read_item(
    item_id: str = Field(description='ID of the item to read'),
):
    # TODO 단계 2-1: items dict에서 item_id 조회
    if item_id not in items:
        raise ValueError(f'Item with id {item_id} not found')
    return items[item_id]

# Smoke test (placeholder data)
items['demo-1'] = {'text': 'Buy groceries', 'done': False}
print('read_item demo:', read_item('demo-1'))
```

## §5. 도구 슬롯 2 — 추가 패턴

두 번째 도구 슬롯은 추가 패턴이다. 함수 이름은 자기 도메인의 항목을 추가하는 의미를 담아 짓고, 새로운 항목을 추가하여 상태를 변경하는 역할을 맡긴다. 항목의 아이디는 함수가 자동으로 생성하거나 입력으로 받는 두 가지 방식 모두 가능하다. 본 예제에서는 입력으로 받는 방식을 사용한다. 자동 생성 방식은 별도의 식별자 생성 로직이 필요하므로 입력 방식이 학습용으로는 더 단순하다.

이 도구는 상태 변경을 일으키므로, 추가하기 전에 중복 아이디 검사 같은 안전 장치를 넣어 두는 것이 좋다. 그렇게 하지 않으면 같은 아이디로 두 번 호출했을 때 기존 데이터가 덮어 씌워지는 위험한 상황이 생길 수 있다. 이런 사소한 부분이 실제 운영에서는 데이터 무결성을 좌우하는 결정적 요소가 되곤 한다.


```python
# Week_07.md §2.6 line ~1467 — Tool 2: add 패턴

@mcp.tool(
    name='add_item',
    description='Add a new item with the given text',
)
def add_item(
    item_id: str = Field(description='Unique ID for the new item'),
    text: str = Field(description='Body text of the item'),
):
    # TODO 단계 2-2: 중복 ID 체크 후 items dict 에 저장
    if item_id in items:
        raise ValueError(f'Item id {item_id} already exists')
    items[item_id] = {'text': text, 'done': False}
    return f'Added item {item_id}'

print(add_item('demo-2', 'Read MCP docs'))
print('items now:', items)
```

## §6. 도구 슬롯 3 — 갱신 패턴

세 번째 도구 슬롯은 갱신 패턴이다. 함수 이름은 자기 도메인의 항목을 갱신하는 의미를 담아 짓고, 이미 존재하는 항목의 필드를 수정하는 역할을 맡긴다. 이 슬롯의 이름은 자기 도메인에 맞춰 변경할 수 있다. 예를 들어 할 일 도메인에서는 할 일 완료 처리라는 이름이 자연스럽고, 레시피 도메인에서는 레시피 평점 매기기가 어울린다. 개인 지식 베이스 도메인이라면 노트 갱신이 적합한 이름이 된다.

갱신 도구도 상태 변경을 일으키므로, 우선 항목의 존재 여부를 확인한 다음에 필드를 수정해야 한다. 존재하지 않는 항목을 갱신하려고 하면 명확한 오류 메시지를 던지는 것이 좋은 설계다. 그래야 사용 측에서 무엇이 잘못되었는지를 빠르게 파악할 수 있다.


```python
# Week_07.md §2.6 line ~1467 — Tool 3: update 패턴

@mcp.tool(
    name='update_item',
    description='Update fields of an existing item',
)
def update_item(
    item_id: str = Field(description='ID of the item to update'),
    text: str = Field(description='New text body (optional)', default=''),
    done: bool = Field(description='Mark complete?', default=False),
):
    # TODO 단계 2-3: 존재 확인 후 필드 갱신
    if item_id not in items:
        raise ValueError(f'Item with id {item_id} not found')
    if text:
        items[item_id]['text'] = text
    items[item_id]['done'] = done
    return f'Updated {item_id} -> {items[item_id]}'

print(update_item('demo-1', text='Buy groceries (urgent)', done=True))
```

## §8. 리소스 슬롯 2 — 매개변수 유알아이 패턴

두 번째 리소스 슬롯은 매개변수 유알아이 패턴이다. 유알아이 안에 매개변수 자리표시자가 들어가 있고, 그 매개변수가 함수 인자로 자동 매핑되는 형태다. 유알아이의 형태는 자기 스킴에 슬래시 두 개와 단일 항목을 의미하는 단어와 자리표시자를 붙이는 식으로 짓는 것이 자연스럽다. 마임 타입은 응답 형태에 맞게 정한다. 단순한 텍스트라면 평문 형태가 적합하고, 객체를 반환한다면 자바스크립트 객체 표기법이 더 적합하다.

앞서 §3 에서 본 것처럼, 유알아이의 자리표시자 이름과 함수 매개변수 이름이 정확히 일치해야 에스디케이가 자동 매핑을 수행할 수 있다는 점을 잊지 말자. 이름이 다르면 매핑이 실패하므로 함수 호출 자체가 일어나지 않는다.

## §9. 프롬프트 슬롯 — 요약 프롬프트

마지막 슬롯은 프롬프트다. 자기 도메인에 어울리는 검증된 요약 템플릿을 작성한다. 본문 안에서 어떤 도구나 리소스를 사용할지를 명시하면, 도구·리소스·프롬프트의 3자 연계가 자연스럽게 만들어진다. 즉 프롬프트가 단순한 지시문이 아니라, 다른 컴포넌트의 사용을 유도하는 안내서 역할을 하게 된다는 뜻이다.

**도메인별 프롬프트 변형 예시**:

할 일 도메인의 프롬프트라면 다음과 같은 식으로 만들 수 있다. 오늘 할 일 중에서 완료 항목과 미완료 항목을 분리하여 요약하라는 지시문이 자연스럽다. 레시피 도메인이라면 이번 주 식단을 균형 있게 제안하라는 식으로, 탄수화물과 단백질과 채소의 비율을 고려하라는 지시를 함께 담을 수 있다. 개인 지식 베이스라면 지난 주 노트들의 핵심 키워드 다섯 개와 액션 아이템을 추출하라는 지시가 자연스럽다.

프롬프트 본문에 도구나 리소스의 이름을 명시적으로 적어 두면, 클로드가 그 지시에 따라 자연스럽게 그 컴포넌트를 호출하게 된다. 이것이 바로 강의에서 반복적으로 강조한 3자 연계의 핵심이다. 즉 프롬프트가 도메인 지식을 담은 안내서가 되어 도구와 리소스의 사용을 조율하는 역할을 하는 것이다.


```python
# Week_07.md §2.6 line ~1469 — Prompt: summarize

@mcp.prompt(
    name='summarize',
    description='Summarize all items into 3 bullet points',
)
def summarize_items() -> list[base.Message]:
    # TODO 단계 4: 도메인에 맞는 검증된 요약 템플릿
    prompt = '''
    Your goal is to summarize the user's items.

    Steps:
    1. Use the items://list resource to find all available item IDs.
    2. For each item, use items://item/{item_id} to fetch its content.
    3. Produce exactly 3 bullet points covering the most important themes.

    Format the output as a markdown bullet list.
    '''
    return [base.UserMessage(prompt)]

msgs = summarize_items()
print('summarize prompt has', len(msgs), 'message(s)')
print('Preview:', str(msgs[0].content)[:120])
```

## §10. 자가 검증 — 클라이언트로 전체 사이클 확인

이제까지 작성한 모든 셀의 내용을 단일 파일인 자기 서버 파일에 한 번에 저장한 다음, 클라이언트로 연결하여 다음 여섯 가지를 차례로 검증한다.

첫째 검증은 도구 목록 조회다. 호출 결과가 세 개의 도구(읽기, 추가, 갱신 패턴)를 반환해야 한다. 둘째 검증은 리소스 목록 조회다. 호출 결과가 직접 유알아이 리소스 한 개를 반환해야 한다. 셋째 검증은 리소스 템플릿 목록 조회다. 호출 결과가 매개변수 유알아이 리소스 한 개를 반환해야 한다. 넷째 검증은 프롬프트 목록 조회다. 호출 결과가 한 개의 프롬프트(요약 또는 자유 주제)를 반환해야 한다. 다섯째 검증은 직접 리소스 읽기다. 호출 결과가 빈 리스트나 임시 아이디들을 반환해야 한다. 여섯째 검증은 통합 동작 검증이다. 도구로 항목을 추가한 다음 다시 리소스를 조회하여 그 항목이 잘 반영되었는지 확인한다.

이 모든 검증이 통과하면 도전 과제가 완료된 것으로 간주한다. 만약 어느 한 가지라도 실패한다면, 해당 컴포넌트의 정의로 돌아가서 무엇이 빠졌는지를 점검해 본다.


```python
# Week_07.md §2.6 line ~1470 — my_mcp_server.py 한 번에 저장

server_code = '''from pydantic import Field
from mcp.server.fastmcp import FastMCP
from mcp.server.fastmcp.prompts import base

mcp = FastMCP("MyDomainMCP", log_level="ERROR")

items: dict = {}


@mcp.tool(name="read_item", description="Read a single item by ID")
def read_item(item_id: str = Field(description="ID of the item to read")):
    if item_id not in items:
        raise ValueError(f"Item with id {item_id} not found")
    return items[item_id]


@mcp.tool(name="add_item", description="Add a new item with the given text")
def add_item(
    item_id: str = Field(description="Unique ID for the new item"),
    text: str = Field(description="Body text of the item"),
):
    if item_id in items:
        raise ValueError(f"Item id {item_id} already exists")
    items[item_id] = {"text": text, "done": False}
    return f"Added item {item_id}"


@mcp.tool(name="update_item", description="Update fields of an existing item")
def update_item(
    item_id: str = Field(description="ID of the item to update"),
    text: str = Field(description="New text body (optional)", default=""),
    done: bool = Field(description="Mark complete?", default=False),
):
    if item_id not in items:
        raise ValueError(f"Item with id {item_id} not found")
    if text:
        items[item_id]["text"] = text
    items[item_id]["done"] = done
    return f"Updated {item_id}"


@mcp.resource("items://list", mime_type="application/json")
def list_items() -> list[str]:
    return list(items.keys())


@mcp.resource("items://item/{item_id}", mime_type="application/json")
def fetch_item(item_id: str) -> dict:
    if item_id not in items:
        raise ValueError(f"Item with id {item_id} not found")
    return items[item_id]


@mcp.prompt(name="summarize", description="Summarize all items into 3 bullet points")
def summarize_items() -> list[base.Message]:
    prompt = """
    Your goal is to summarize the user\'s items.

    Steps:
    1. Use the items://list resource to find all available item IDs.
    2. For each item, use items://item/{item_id} to fetch its content.
    3. Produce exactly 3 bullet points covering the most important themes.
    """
    return [base.UserMessage(prompt)]


if __name__ == "__main__":
    mcp.run(transport="stdio")
'''

with open('my_mcp_server.py', 'w', encoding='utf-8') as f:
    f.write(server_code)
print('my_mcp_server.py saved.')
```

```python
# Week_07.md §2.6 line ~1470 — MCPClient 자가 검증
import sys
if 'mcp_client' in sys.modules:
    del sys.modules['mcp_client']
from mcp_client import MCPClient
import os

USE_UV = os.getenv('USE_UV', '0') == '1'
command, args = ('uv', ['run', 'my_mcp_server.py']) if USE_UV else ('python', ['my_mcp_server.py'])

async def self_verify():
    async with MCPClient(command=command, args=args) as client:
        # 1. Tools
        tools = await client.list_tools()
        print(f'[1] Tools ({len(tools)}):', [t.name for t in tools])
        assert len(tools) == 3, f'Expected 3 tools, got {len(tools)}'

        # 2. Prompts
        prompts = await client.list_prompts()
        print(f'[2] Prompts ({len(prompts)}):', [p.name for p in prompts])
        assert len(prompts) >= 1, f'Expected ≥1 prompt, got {len(prompts)}'

        # 3. Direct resource
        item_ids = await client.read_resource('items://list')
        print(f'[3] items://list -> {item_ids}')

        # 4. Add an item via tool, then verify via resource
        await client.call_tool('add_item', {'item_id': 'task-1', 'text': 'Verify MCP server'})
        item_ids2 = await client.read_resource('items://list')
        print(f'[4] After add_item -> items://list = {item_ids2}')
        assert 'task-1' in item_ids2, 'add_item or read_resource failed'

        # 5. Templated resource
        item = await client.read_resource('items://item/task-1')
        print(f'[5] items://item/task-1 -> {item}')

        # 6. Prompt fetch
        msgs = await client.get_prompt('summarize', {})
        print(f'[6] /summarize -> {len(msgs)} message(s)')

        print('\n✅ All checks passed — challenge complete!')
        return tools, prompts

await self_verify()
```
