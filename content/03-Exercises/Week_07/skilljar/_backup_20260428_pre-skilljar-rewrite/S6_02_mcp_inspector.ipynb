## §1. MCP Inspector 실행

MCP Inspector는 브라우저 기반 테스트 도구입니다.  
`mcp dev` 명령으로 서버를 Inspector에 연결하여 테스트합니다.

### 터미널에서 실행
```bash
# S6_01에서 생성한 server.py를 Inspector로 테스트
mcp dev server.py
```

브라우저에서 Inspector UI가 열리면:
1. **Tools 탭**: 등록된 도구 목록 확인 + 실행 테스트
2. **Resources 탭**: 리소스 목록 (아직 없음)
3. **Prompts 탭**: 프롬프트 목록 (아직 없음)
4. **Log 탭**: JSON-RPC 메시지 실시간 모니터링

````python
# 노트북에서 Inspector를 백그라운드로 실행 (선택사항)
# 보통은 터미널에서 직접 실행하는 것이 더 편합니다.
import subprocess
import os

# server.py가 존재하는지 확인
if os.path.exists("server.py"):
    print("server.py를 찾았습니다.")
    print("\n터미널에서 다음 명령을 실행하세요:")
    print("  mcp dev server.py")
    print("\n브라우저에서 Inspector UI가 열립니다.")
else:
    print("server.py를 찾을 수 없습니다.")
    print("먼저 S6_01_mcp_server.ipynb를 실행하여 server.py를 생성하세요.")
````

## §2. Inspector에서 도구 테스트하기

Inspector의 **Tools 탭**에서 다음을 수행합니다:

### 테스트 1: get_current_time
```
Tool: get_current_time
Parameters:
  format: "%Y-%m-%d %H:%M"
[Run] → Result: "2026-04-13 14:30"
```

### 테스트 2: add_numbers
```
Tool: add_numbers
Parameters:
  a: 3.14
  b: 2.71
[Run] → Result: 5.85
```

### 테스트 3: calculate_area
```
Tool: calculate_area
Parameters:
  width: 300
  height: 600
  unit: "mm"
[Run] → Result: "180000.00 mm²"
```

> **Log 탭**에서 JSON-RPC 메시지를 확인하면 클라이언트-서버 통신을 직접 볼 수 있습니다.

## §3. Claude Desktop에 서버 연결

Inspector 테스트가 완료되면, Claude Desktop에 서버를 연결합니다.

```bash
# Claude Desktop에 서버 설치
mcp install server.py
```

이 명령은 `claude_desktop_config.json`에 서버를 자동 등록합니다:

```json
{
    "mcpServers": {
        "My First MCP Server": {
            "command": "python",
            "args": ["/absolute/path/to/server.py"]
        }
    }
}
```

Claude Desktop을 재시작하면 서버의 도구를 사용할 수 있습니다.

## §4. Claude Code에 서버 연결

Claude Code에서도 MCP 서버를 연결할 수 있습니다.

```bash
# Claude Code의 프로젝트 설정에 추가
claude mcp add my-server -- python /absolute/path/to/server.py
```

또는 `.claude/settings.json`에 직접 추가:

```json
{
    "mcpServers": {
        "my-server": {
            "command": "python",
            "args": ["/absolute/path/to/server.py"]
        }
    }
}
```

````python
# 현재 server.py의 절대 경로 출력 (연결 시 필요)
import os

server_path = os.path.abspath("server.py")
print(f"서버 절대 경로: {server_path}")
print(f"\nClaude Desktop 설치: mcp install {server_path}")
print(f"Claude Code 추가:   claude mcp add my-server -- python {server_path}")
````

## 핵심 정리

| 명령 | 용도 |
|------|------|
| `mcp dev server.py` | Inspector로 서버 테스트 (개발용) |
| `mcp install server.py` | Claude Desktop에 서버 설치 |
| `claude mcp add` | Claude Code에 서버 추가 |

**개발 워크플로**: 코드 수정 → `mcp dev` 테스트 → `mcp install` 배포
