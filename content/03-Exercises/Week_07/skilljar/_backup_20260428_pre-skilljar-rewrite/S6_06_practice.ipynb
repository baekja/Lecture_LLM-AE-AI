## Step 4: 테스트

등록된 도구, 리소스, 프롬프트를 확인하고 테스트하세요.

```python
import asyncio

async def test_my_server():
    # 도구 목록
    tools = await mcp.list_tools()
    print(f"도구 ({len(tools)}개):")
    for t in tools:
        print(f"  - {t.name}: {t.description}")

    # 리소스 목록
    resources = await mcp.list_resources()
    print(f"\n리소스 ({len(resources)}개):")
    for r in resources:
        print(f"  - {r.uri}: {r.name}")

    # 프롬프트 목록
    prompts = await mcp.list_prompts()
    print(f"\n프롬프트 ({len(prompts)}개):")
    for p in prompts:
        print(f"  - {p.name}: {p.description}")

    # TODO: 도구 호출 테스트
    # result = await mcp.call_tool("my_tool_1", {...})
    # print(f"\n도구 결과: {result}")

await test_my_server()
```
