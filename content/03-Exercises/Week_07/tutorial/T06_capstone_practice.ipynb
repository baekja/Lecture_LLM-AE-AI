동작하는 placeholder. 자신의 도메인에 맞게 함수 이름·로직을 수정.

```python
@mcp.tool(name="read_item", description="아이템 ID로 단일 아이템 조회")
def read_item(item_id: str = Field(description="아이템 ID")) -> str:
    if item_id not in items:
        raise ValueError(f"Item {item_id} not found")
    return items[item_id]

# TODO: 자신의 도메인에 맞춰 함수 이름·인자·로직 변경
#   예) read_recipe(recipe_id), read_workout(workout_id)

print("도구 1 등록 완료")
```

## §7. 리소스 1 — direct (`<scheme>://list`)

```python
@mcp.resource("items://list", mime_type="application/json")
def list_items() -> str:
    """모든 아이템 ID 목록."""
    return json.dumps(list(items.keys()), ensure_ascii=False)

# TODO: 자신의 도메인 URI 스킴으로 변경
#   예) recipes://list, gym://history

print("리소스 1 (direct) 등록 완료")
```

## §8. 리소스 2 — templated (`<scheme>://item/{id}`)

```python
@mcp.resource("items://item/{item_id}", mime_type="text/plain")
def fetch_item(item_id: str) -> str:
    """개별 아이템 조회."""
    return items.get(item_id, f"Not found: {item_id}")

# TODO: 자신의 도메인 URI 템플릿으로 변경
#   예) recipes://recipe/{recipe_id}, gym://workout/{workout_id}

print("리소스 2 (templated) 등록 완료")
```

## §9. 프롬프트 1 — `summarize`

```python
@mcp.prompt(name="summarize", description="아이템 요약")
def summarize_items() -> list[base.Message]:
    """전체 아이템을 카테고리별 요약."""
    ids = list(items.keys())
    if not ids:
        prompt = "아이템이 없습니다. 먼저 add_item으로 데이터를 추가하세요."
    else:
        sample = "\n".join(f"- {k}: {items[k][:50]}..." for k in ids[:10])
        prompt = f"다음 {len(ids)}개 아이템을 카테고리별로 요약해주세요:\n{sample}"
    return [base.UserMessage(prompt)]

# TODO: 자신의 도메인 맞춤 프롬프트로 확장
#   예) weekly_review (개인 KB), meal_plan (레시피), progress_report (운동)

print("프롬프트 1 등록 완료")
```

> ☑ **체크포인트 3**: 프롬프트 1개 등록
>
> `await mcp.list_prompts()`로 1개 표시 확인.

## §10. 자가 검증 — 전체 등록 확인

도구 / 리소스 / 프롬프트 모두 정상 등록되었는지 확인.

```python
import asyncio

async def validate_capstone():
    tools = await mcp.list_tools()
    resources = await mcp.list_resources()
    templates = await mcp.list_resource_templates()
    prompts = await mcp.list_prompts()

    print(f"=== Capstone 등록 현황 ===")
    print(f"Tools: {len(tools)}개 (목표: 3)")
    for t in tools:
        print(f"  - {t.name}")
    print(f"Resources (direct): {len(resources)}개 (목표: 1)")
    for r in resources:
        print(f"  - {r.uri}")
    print(f"Resources (templated): {len(templates)}개 (목표: 1)")
    for t in templates:
        print(f"  - {t.uriTemplate}")
    print(f"Prompts: {len(prompts)}개 (목표: 1)")
    for p in prompts:
        print(f"  - {p.name}")

    # 통과 기준
    pass_check = (len(tools) >= 3 and len(resources) >= 1
                  and len(templates) >= 1 and len(prompts) >= 1)
    print(f"\n{'✅ 통과' if pass_check else '❌ 미달'}")

await validate_capstone()
```
